```python
import pandas as pd
import plotly.express as px
import plotly.graph_objects as go

# Load the Gapminder dataset
df = px.data.gapminder()

# Display basic information
print("Dataset loaded successfully!")
print("Rows and columns:", df.shape)
print("\nColumns:")
print(df.columns.tolist())

df.head()
```

```python
# Check the years and countries in the dataset
print("Number of countries:", df["country"].nunique())
print("Continents:", df["continent"].unique())
print("Year range:", df["year"].min(), "-", df["year"].max())

# Check the main numerical columns
df[["lifeExp", "pop", "gdpPercap"]].describe()
```

```python
# Select data for 2007
df_2007 = df[df["year"] == 2007]

# Create a scatter plot
fig = px.scatter(
    df_2007,
    x="gdpPercap",
    y="lifeExp",
    color="continent",
    hover_name="country",
    title="Higher GDP per Capita Is Generally Associated with Longer Life Expectancy (2007)",
    labels={
        "gdpPercap": "GDP per Capita",
        "lifeExp": "Life Expectancy (Years)",
        "continent": "Continent"
    }
)

fig.update_layout(
    template="plotly_white",
    showlegend=True,
    xaxis=dict(showgrid=False, showline=False, zeroline=False),
    yaxis=dict(showgrid=False, showline=False, zeroline=False)
)

fig.show()
```

```python
# Bubble chart: wealth, life expectancy, and population
fig = px.scatter(
    df_2007,
    x="gdpPercap",
    y="lifeExp",
    size="pop",
    color="continent",
    hover_name="country",
    size_max=55,
    title="Country Wealth and Life Expectancy, with Population Shown by Bubble Size (2007)",
    labels={
        "gdpPercap": "GDP per Capita",
        "lifeExp": "Life Expectancy (Years)",
        "pop": "Population",
        "continent": "Continent"
    }
)

fig.update_layout(
    template="plotly_white",
    xaxis=dict(showgrid=False, showline=False, zeroline=False),
    yaxis=dict(showgrid=False, showline=False, zeroline=False)
)

fig.show()
```

# Key Findings and Conclusion

## Task 1: GDP and Life Expectancy

The 2007 scatter plot shows a general positive relationship between GDP per capita and life expectancy. Countries with higher GDP per capita tend to have longer life expectancy, although there are exceptions. African countries are often clustered at lower GDP per capita and lower life expectancy.

## Task 2: Bubble Chart and Population

The bubble chart adds population as a third variable. Countries with larger populations appear as larger bubbles, making it easier to compare population alongside GDP per capita and life expectancy. The chart shows that countries differ considerably in wealth, population, and health outcomes.

## Overall Conclusion

The scatter plot and bubble chart reveal patterns in the 2007 Gapminder data. GDP per capita is generally associated with life expectancy, but the relationship is not perfect. Population adds further context to the comparison. These charts show associations, not proof that higher GDP directly causes longer life expectancy.
